```python

import random
import math

# 1. COST FUNCTION
def calculate_cost(board):
    cost = 0

    for i in range(8):
        for j in range(i + 1, 8):

            if board[i] == board[j]:
                cost += 1

            elif abs(board[i] - board[j]) == abs(i - j):
                cost += 1

    return cost


# 2. GENERATE RANDOM NEIGHBOUR
def generate_neighbour(board):
    neighbour = board.copy()

    column = random.randint(0, 7)
    new_row = random.randint(0, 7)

    while new_row == neighbour[column]:
        new_row = random.randint(0, 7)

    neighbour[column] = new_row

    return neighbour


# 3. DISPLAY CHESS BOARD
def display_board(board):
    for row in range(8):
        for col in range(8):

            if board[col] == row:
                print("Q", end=" ")
            else:
                print(".", end=" ")

        print()


# 4. DISPLAY LEVEL
def display_level(level, board):
    print("\n==========================================")
    print(level)
    print("==========================================")

    print("Board :", board)
    print("Cost  :", calculate_cost(board))

    display_board(board)


# 5. SIMULATED ANNEALING
def simulated_annealing(initial_board, initial_temperature,
                        cooling_factor, max_iterations):

    current = initial_board.copy()
    current_cost = calculate_cost(current)

    best = current.copy()
    best_cost = current_cost

    T = initial_temperature
    iteration = 0

    MIN_TEMPERATURE = 0.000001

    # Store intermediate solutions
    levels = {}

    level1 = max(1, max_iterations // 3)
    level2 = max(2, (2 * max_iterations) // 3)

    while T > MIN_TEMPERATURE and iteration < max_iterations:

        iteration += 1

        next_state = generate_neighbour(current)
        next_cost = calculate_cost(next_state)

        delta_E = current_cost - next_cost

        if delta_E > 0:
            current = next_state
            current_cost = next_cost

        else:
            probability = math.exp(delta_E / T)
            random_number = random.random()

            if random_number < probability:
                current = next_state
                current_cost = next_cost

        # Update best solution
        if current_cost < best_cost:
            best = current.copy()
            best_cost = current_cost

        # Save intermediate levels
        if iteration == level1:
            levels[1] = best.copy()

        if iteration == level2:
            levels[2] = best.copy()

        # Cool down
        T = T * cooling_factor

    # If temperature becomes too low before a checkpoint,
    # use the best solution obtained so far.
    if 1 not in levels:
        levels[1] = best.copy()

    if 2 not in levels:
        levels[2] = best.copy()

    return best, best_cost, iteration, levels


# 6. MAIN PROGRAM
print("==========================================")
print("       8-QUEENS SIMULATED ANNEALING")
print("==========================================")

print("\nBoard Representation:")
print("board[column] = row")
print("Rows and columns are numbered from 0 to 7.")

# INPUT INITIAL BOARD
while True:
    try:
        board = list(map(int, input(
            "\nEnter initial board (8 values): "
        ).split()))

        if len(board) != 8:
            print("Enter exactly 8 values.")
            continue

        if any(row < 0 or row > 7 for row in board):
            print("Values must be between 0 and 7.")
            continue

        break

    except ValueError:
        print("Enter integers only.")


# INPUT TEMPERATURE
while True:
    try:
        temperature = float(input(
            "Enter initial temperature (e.g. 10): "
        ))

        if temperature <= 0:
            print("Temperature must be greater than 0.")
            continue

        break

    except ValueError:
        print("Enter a valid number.")


# INPUT COOLING FACTOR
while True:
    try:
        alpha = float(input(
            "Enter cooling factor (e.g. 0.99): "
        ))

        if alpha <= 0 or alpha >= 1:
            print("Cooling factor must be between 0 and 1.")
            continue

        break

    except ValueError:
        print("Enter a valid number.")


# INPUT MAXIMUM ITERATIONS
while True:
    try:
        max_iterations = int(input(
            "Enter maximum iterations (e.g. 100000): "
        ))

        if max_iterations <= 0:
            print("Iterations must be greater than 0.")
            continue

        break

    except ValueError:
        print("Enter a valid integer.")


# INITIAL STATE
print("\n==========================================")
print("INITIAL STATE")
print("==========================================")

print("Initial Board :", board)
print("Initial Cost  :", calculate_cost(board))
print("Temperature   :", temperature)
print("Cooling Factor:", alpha)

display_board(board)


# RUN SIMULATED ANNEALING
solution, final_cost, iterations, levels = simulated_annealing(
    board,
    temperature,
    alpha,
    max_iterations
)


# DISPLAY INTERMEDIATE LEVELS
display_level("LEVEL 1 SOLUTION", levels[1])

display_level("LEVEL 2 SOLUTION", levels[2])


# FINAL RESULT
print("\n==========================================")
print("FINAL RESULT")
print("==========================================")

print("Final Board :", solution)
print("Final Cost  :", final_cost)
print("Iterations  :", iterations)

print("\n==========================================")
print("FINAL OUTPUT STATE")
print("==========================================")

display_board(solution)

print("\nFinal Cost:", final_cost)

if final_cost == 0:
    print("SUCCESS: 8-Queens solution found!")
else:
    print("No solution found within the iteration limit.")
```
